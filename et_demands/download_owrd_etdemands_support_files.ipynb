Establish an SSH connection with the DRI FTP Server

```python
ssh_client = paramiko.SSHClient()
ssh_client.set_missing_host_key_policy(paramiko.AutoAddPolicy()) # Automatically add host keys
ssh_client.connect(
    hostname='crushftp.dri.edu',
    port=22, # Default SFTP port
    username='owrd-etdownload', #read-only access user
    password='8xCuMjvCK2WdGlT' 
)
```

Open an SFTP session

```python
sftp_client = ssh_client.open_sftp()

# If is_closed is True, the connection is closed; otherwise, it's open.
if ~sftp_client.sock.closed:
    print('Connection successful and open.')
else:
    print('Connection failed. Try again or check hosename and credentials')
```

Download historical gridMET weather data

```python
remote_gridmet_zip_path = 'owrd_historical_gridmet_1979_2024.zip'
local_gridmet_zip_path = r'D:\owrd_etdemands\owrd_historical_gridmet_1979_2024.zip'
sftp_client.get(remote_gridmet_zip_path, local_gridmet_zip_path)
print(f"File downloaded from {remote_gridmet_zip_path} to {local_gridmet_zip_path}")
```

Download original OWRD ET Demands Support Files

```python
remote_etd_zip_path = 'owrd_historical_etdemands_files.zip'
local_etd_zip_path = r'D:\owrd_etdemands\owrd_historical_etdemands_files.zip'
sftp_client.get(remote_etd_zip_path, local_etd_zip_path)
print(f"File downloaded from {remote_etd_zip_path} to {local_etd_zip_path}")
```

Close the SFTP Connection

```python
# Close the SFTP session
sftp_client.close()

# Close the underlying SSH connection
ssh_client.close()
```
