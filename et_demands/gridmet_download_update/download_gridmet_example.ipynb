## GridMET timeseries download example
The following jupyter notebook shows how to download or update gridmet climate data using the 'download_gridmet_opendap.py' function. The notebook constructs a download command based on user specified inputs (year, folder_path, etc.) and executes the script to download gridmet climate time series from the https://www.opendap.org Thredds server. Reference ET (ETo and ETr) is computed using the python refet package (https://github.com/WSWUP/RefET).

'download_gridmet_opendap.py' can also be run directly from the command line if desired. For example: 
python C:\dri-owrd-et\et_demands\gridmet_download_update\download_gridmet_opendap.py -i C:\dri-owrd-et\et_demands\gridmet_download_update\or_et_cells_pts_example.csv -o C:\dri-owrd-et\et_demands\gridmet_download_update\climate_data --years 1979-2024

Access the download script help information by opening the script in a text editor or running the following for input agrument options:
python C:\dri-owrd-et\et_demands\gridmet_download_update\download_gridmet_opendap.py -h

```python
#specify the year range of gridmet weather data you'd like to download. years are inclusive.
start_yr = 1979
end_yr = 2024

# location of the gridmet download python script
function_path = 'C:\dri-owrd-et\et_demands\gridmet_download_update\download_gridmet_opendap.py'

# location of the gridmet cell list
# input_csv = 'C:\dri-owrd-et\et_demands\or_model_files\or_et_cells_pts_example.csv'
input_csv = 'C:\dri-owrd-et\et_demands\or_model_files\or_et_cells_pts_all.csv'

# location of the folder you'd like to save the climate timeseries data
out_folder = 'D:\owrd_etdemands\owrd_historical_gridmet_1979_2024'

# flag to overwrite existing data/years if climate data has been previsouly downloaded
update_flag = False
```

```python
# Bulid python command based on parameters listed above
python_cmd = '{} -i {} -o {} --years {}-{}'.format(function_path, input_csv, out_folder, start_yr, end_yr)

if update_flag:
    python_cmd = python_cmd+ ' --update'
else:
    None

print(python_cmd)
```
