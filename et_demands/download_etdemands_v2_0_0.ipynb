The OWRD ET Demands model was set-up and run using ET Demands v2.0.0. The following notebook retrieves ET Demands v2.0.0 and saves it to a user specified directory. 

```python
# import python packages
import os
import requests
import zipfile
```

```python
# specify directory to save et demands model files
save_dir = r'C:\dri-owrd-et'

if os.path.isdir(save_dir):
    print(f"The directory '{save_dir}' exists. Continue to download ET Demands model code.")
else:
    print(f"The directory '{save_dir}' does not exist. Check folders before proceeding.")

```

```python
# ET Demands v2.0.0 Release .zip URL
URL = 'https://github.com/WSWUP/et-demands/archive/refs/tags/v2.0.0.zip'

# Try to get the ZIP file
try:
    response = requests.get(URL)
except OSError:
    print('No connection to the server!')

# check if the request is succesful
if response.status_code == 200:
    # Save dataset to file
    print('Status 200, OK. Downloading file.')
    open(os.path.join(save_dir, 'et_demands_v2_0_0.zip'), 'wb').write(response.content)
else:
    print('ZIP file request not successful!.')
```

```python
# Unzip
if os.path.isdir(os.path.join(save_dir, 'et-demands-2.0.0')):
    print('ET Demands model code directory already exists. Check folders before proceeding.')
else:
    with zipfile.ZipFile(os.path.join(save_dir, 'et_demands_v2_0_0.zip'), 'r') as zip_ref:
        zip_ref.extractall(save_dir)
        print('Unzipped ET Demands model v2.0.0 files saved to: {}{} .'.format(save_dir, '\et-demands-2.0.0'))
```
