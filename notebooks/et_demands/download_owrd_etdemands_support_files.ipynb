The following script retrieves OR ET Demands model and gridMET climate data files. Files are stored on a project specific SFTP server hosted by DRI. 

```python
import os
import paramiko
import zipfile
import lzma
import pyzipper
```

```python
save_dir = r'C:\dri-owrd-et\owrd_etdemands'

try:
    os.makedirs(save_dir, exist_ok=False)
    print(f"Directory '{save_dir}' created.")
except OSError:
    print(f"Directory '{save_dir}' already exists. Check folders before proceeding.")
```

Establish an SSH connection with the DRI FTP Server

```python
ssh_client = paramiko.SSHClient()
ssh_client.set_missing_host_key_policy(paramiko.AutoAddPolicy()) # Automatically add host keys
ssh_client.connect(
    hostname='crushftp.dri.edu',
    port=22, # Default SFTP port
    username='owrd-etdownload', #read-only access user
    password='8xCuMjvCK2WdGlT' 
)
```

Open an SFTP session

```python
sftp_client = ssh_client.open_sftp()

# If is_closed is True, the connection is closed; otherwise, it's open.
if ~sftp_client.sock.closed:
    print('Connection successful and open.')
else:
    print('Connection failed. Try again or check hosename and credentials')
```

Download historical gridMET weather data package. This .zip contains previously download gridMET .csv files used for ET Demands model application.

This download is ~4GB of data. 

```python
remote_gridmet_zip_path = 'owrd_historical_gridmet_1979_2024.zip'
local_gridmet_zip_path = r'{}\owrd_historical_gridmet_1979_2024.zip'.format(save_dir)
sftp_client.get(remote_gridmet_zip_path, local_gridmet_zip_path)
print(f"File downloaded from {remote_gridmet_zip_path} to {local_gridmet_zip_path}")
```

Download historical OWRD ET Demands support files. This .zip contains et demands model files used for historical (1980-2022) model application. 

```python
remote_etd_hist_zip_path = 'owrd_etdemands_historical.zip'
local_etd_hist_zip_path = r'{}\owrd_etdemands_historical.zip'.format(save_dir)
sftp_client.get(remote_etd_hist_zip_path, local_etd_hist_zip_path)
print(f"File downloaded from {remote_etd_hist_zip_path} to {local_etd_hist_zip_path}")
```

Download 2023-2024 update OWRD ET Demands support files. This .zip contains files used to support et demands updates.

```python
remote_etd_zip_path = 'owrd_etdemands_update.zip'
local_etd_zip_path = r'{}\owrd_etdemands_update.zip'.format(save_dir)
sftp_client.get(remote_etd_zip_path, local_etd_zip_path)
print(f"File downloaded from {remote_etd_zip_path} to {local_etd_zip_path}")
```

Close the SFTP Connection

```python
# Close the SFTP session
sftp_client.close()

# Close the underlying SSH connection
ssh_client.close()
```

```python
# Unzip ET Demands historical model files.
with zipfile.ZipFile(local_etd_hist_zip_path, "r") as zip_ref:
    zip_ref.extractall(save_dir)
    print('Unzipped ET Demands historical model files to:  {}.'.format(save_dir))
```

```python
# Unzip ET Demands update model files. 
with zipfile.ZipFile(local_etd_zip_path, "r") as zip_ref:
    zip_ref.extractall(save_dir)
    print('Unzipped ET Demands update model files to:  {}.'.format(save_dir))
```

```python
# Unzip gridMET data files.
with zipfile.ZipFile(local_gridmet_zip_path, "r") as zip_ref:
    zip_ref.extractall(save_dir)
    print('Unzipped gridMET .csv files to:  {}.'.format(save_dir))
```
